### Вариант задания 1


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.

Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IPerson
{
    string Name { get; set; }
    int Age { get; set; }
    string GetInfo();
}

public interface IWorker
{
    string Company { get; set; }
    void Work();
}

public class Person : IPerson
{
    private string _name;
    private int _age;
    private string _gender;
    private string _hobby;
    private string _nationality;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set { _age = value; }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public string Hobby
    {
        get { return _hobby; }
        set { _hobby = value; }
    }

    public string Nationality
    {
        get { return _nationality; }
        set { _nationality = value; }
    }

    public Person() { }

    public Person(string name, int age, string gender, string hobby, string nationality)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Hobby = hobby;
        Nationality = nationality;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, {Age} лет, {Gender}, Хобби: {Hobby}, Национальность: {Nationality}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}");
    }

    public virtual void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}.");
    }

    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    public void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }
}

public class Student : Person
{
    private string _university;
    private int _course;
    private string _favoriteSubject;
    private double _averageGrade;

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public int Course
    {
        get { return _course; }
        set { _course = value; }
    }

    public string FavoriteSubject
    {
        get { return _favoriteSubject; }
        set { _favoriteSubject = value; }
    }

    public double AverageGrade
    {
        get { return _averageGrade; }
        set { _averageGrade = value; }
    }

    public Student() : base() { }

    public Student(string name, int age, string gender, string hobby, string nationality,
                   string university, int course, string favoriteSubject, double averageGrade)
        : base(name, age, gender, hobby, nationality)
    {
        University = university;
        Course = course;
        FavoriteSubject = favoriteSubject;
        AverageGrade = averageGrade;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Курс: {Course}, Любимый предмет: {FavoriteSubject}, Средний балл: {AverageGrade}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, учусь на {Course} курсе в {University}, люблю {Hobby}!");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (студент) обсуждает учебу с {other.Name}.");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} учится по предмету {FavoriteSubject}.");
    }

    public void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен в {University}.");
    }
}

public class Employee : Person, IWorker
{
    private string _company;
    private int _salary;
    private string _position;
    private int _experience;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public int Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set { _experience = value; }
    }

    public Employee() : base() { }

    public Employee(string name, int age, string gender, string hobby, string nationality,
                    string company, int salary, string position, int experience)
        : base(name, age, gender, hobby, nationality)
    {
        Company = company;
        Salary = salary;
        Position = position;
        Experience = experience;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, Зарплата: {Salary} руб., Стаж: {Experience} лет";
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (работник) разговаривает с {other.Name}.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} на должности {Position}.");
    }

    public void GetPromotion()
    {
        Salary += 10000;
        Console.WriteLine($"{Name} получил повышение! Новая зарплата: {Salary} руб.");
    }
}

public class Teacher : Employee
{
    private string _subject;
    private string _university;
    private string _academicDegree;
    private int _studentsCount;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public string AcademicDegree
    {
        get { return _academicDegree; }
        set { _academicDegree = value; }
    }

    public int StudentsCount
    {
        get { return _studentsCount; }
        set { _studentsCount = value; }
    }

    public Teacher() : base() { }

    public Teacher(string name, int age, string gender, string hobby, string nationality,
                   string company, int salary, string position, int experience,
                   string subject, string university, string academicDegree, int studentsCount)
        : base(name, age, gender, hobby, nationality, company, salary, position, experience)
    {
        Subject = subject;
        University = university;
        AcademicDegree = academicDegree;
        StudentsCount = studentsCount;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Университет: {University}, Степень: {AcademicDegree}, Студентов: {StudentsCount}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, преподаю {Subject} в {University} уже {Experience} лет! {AcademicDegree}");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (преподаватель) общается с {other.Name}.");
    }

    public void Teach()
    {
        Console.WriteLine($"{Name} преподает {Subject} в {University}.");
    }

    public void CheckHomework()
    {
        Console.WriteLine($"{Name} проверяет домашние задания {StudentsCount} студентов.");
    }
}

public class WorkingStudent : Student, IWorker
{
    private string _company;
    private string _position;
    private int _salary;
    private int _workHours;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public int Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public int WorkHours
    {
        get { return _workHours; }
        set { _workHours = value; }
    }

    public WorkingStudent() : base() { }

    public WorkingStudent(string name, int age, string gender, string hobby, string nationality,
                          string university, int course, string favoriteSubject, double averageGrade,
                          string company, string position, int salary, int workHours)
        : base(name, age, gender, hobby, nationality, university, course, favoriteSubject, averageGrade)
    {
        Company = company;
        Position = position;
        Salary = salary;
        WorkHours = workHours;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, Зарплата: {Salary} руб., Рабочих часов: {WorkHours}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, учусь в {University} и работаю в {Company}.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} на должности {Position}.");
    }
}

Student yulia = new Student("Юля", 19, "Женский", "Роблокс", "Русская", "ТИУ", 2, "Математика", 4.5);
Employee vladimir = new Employee("Владимир", 43, "Мужской", "Автозвук", "Русский", "Газпром", 150000, "Начальник отдела", 15);
Teacher svetlana = new Teacher("Светлана", 47, "Женский", "Чтение", "Русская", "ТИУ", 80000, "Преподаватель", 15, "Математика", "ТИУ", "Кандидат наук", 120);
WorkingStudent maxim = new WorkingStudent("Максим", 22, "Мужской", "Спорт", "Русский", "ТИУ", 4, "Программирование", 4.8, "Яндекс", "Стажер", 40000, 20);
Person sofia = new Person("София", 19, "Женский", "Чтение", "Русская");

Console.WriteLine("=== Информация ===");
Console.WriteLine(yulia.GetInfo());
Console.WriteLine(vladimir.GetInfo());
Console.WriteLine(svetlana.GetInfo());
Console.WriteLine(maxim.GetInfo());
Console.WriteLine(sofia.GetInfo());

Console.WriteLine("\n=== Приветствия ===");
yulia.SayHello();
vladimir.SayHello();
svetlana.SayHello();
maxim.SayHello();
sofia.SayHello();

Console.WriteLine("\n=== Уникальные методы ===");
yulia.Study();
yulia.TakeExam();
vladimir.Work();
vladimir.GetPromotion();
svetlana.Teach();
svetlana.CheckHomework();
maxim.Work();

Console.WriteLine("\n=== Взаимодействие объектов ===");
yulia.InteractWith(svetlana);
svetlana.InteractWith(yulia);
vladimir.InteractWith(yulia);
maxim.InteractWith(svetlana);
```
