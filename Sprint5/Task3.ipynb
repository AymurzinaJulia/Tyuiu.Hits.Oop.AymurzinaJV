----

Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.

Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.
#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
public interface IPerson
{
    string Name { get; set; }
    int Age { get; set; }
    string GetInfo();
    void SayHello();
    string Email { get; set; }
    string Phone { get; set; }
    void SendMessage(string text);
}

public interface IWorker
{
    string Company { get; set; }
    void Work();
    string Department { get; set; }
    void TakeBreak();
}

public interface IStudent
{
    string University { get; set; }
    int Course { get; set; }
    void Study();
    string Group { get; set; }
    void AttendLecture();
}

public class Person : IPerson
{
    private string _name;
    private int _age;
    private string _gender;
    private string _hobby;
    private string _nationality;
    private string _email;
    private string _phone;
    private string _address;
    private string _maritalStatus;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set { _age = value; }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public string Hobby
    {
        get { return _hobby; }
        set { _hobby = value; }
    }

    public string Nationality
    {
        get { return _nationality; }
        set { _nationality = value; }
    }

    public string Email
    {
        get { return _email; }
        set { _email = value; }
    }

    public string Phone
    {
        get { return _phone; }
        set { _phone = value; }
    }

    public string Address
    {
        get { return _address; }
        set { _address = value; }
    }

    public string MaritalStatus
    {
        get { return _maritalStatus; }
        set { _maritalStatus = value; }
    }

    public Person() { }

    public Person(string name, int age, string gender, string hobby, string nationality)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Hobby = hobby;
        Nationality = nationality;
    }

    public Person(string name, int age, string gender, string hobby, string nationality,
                  string email, string phone, string address, string maritalStatus)
        : this(name, age, gender, hobby, nationality)
    {
        Email = email;
        Phone = phone;
        Address = address;
        MaritalStatus = maritalStatus;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, {Age} лет, {Gender}, Хобби: {Hobby}, Национальность: {Nationality}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}");
    }

    public virtual void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}.");
    }

    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    public void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }

    public void Feed(string food)
    {
        Console.WriteLine($"{Name} ест {food}.");
    }

    public void Feed(string food, int amount)
    {
        Console.WriteLine($"{Name} ест {amount} порций {food}.");
    }

    public void Feed(string food, int amount, bool isFavorite)
    {
        Console.WriteLine($"{Name} ест {amount} порций {food}. Любимая еда: {isFavorite}.");
    }

    public void SendMessage(string text)
    {
        Console.WriteLine($"{Name} отправляет сообщение: {text}");
    }

    public void Call(string number)
    {
        Console.WriteLine($"{Name} звонит по номеру {number}.");
    }

    public void UpdateAddress(string newAddress)
    {
        Address = newAddress;
        Console.WriteLine($"{Name} сменил адрес на {newAddress}.");
    }

    public void CelebrateBirthday()
    {
        Age++;
        Console.WriteLine($"{Name} отпраздновал день рождения! Теперь ему {Age} лет.");
    }

    string IPerson.GetInfo()
    {
        return $"{Name}, {Age} лет";
    }

    void IPerson.SayHello()
    {
        Console.WriteLine($"Привет от {Name}");
    }

    string IPerson.Email
    {
        get { return _email; }
        set { _email = value; }
    }

    string IPerson.Phone
    {
        get { return _phone; }
        set { _phone = value; }
    }

    void IPerson.SendMessage(string text)
    {
        Console.WriteLine($"IPerson: {Name} -> {text}");
    }
}

public class Student : Person, IStudent
{
    private string _university;
    private int _course;
    private string _favoriteSubject;
    private double _averageGrade;
    private string _group;
    private string _studentId;
    private int _credits;
    private string _scholarship;

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public int Course
    {
        get { return _course; }
        set { _course = value; }
    }

    public string FavoriteSubject
    {
        get { return _favoriteSubject; }
        set { _favoriteSubject = value; }
    }

    public double AverageGrade
    {
        get { return _averageGrade; }
        set { _averageGrade = value; }
    }

    public string Group
    {
        get { return _group; }
        set { _group = value; }
    }

    public string StudentId
    {
        get { return _studentId; }
        set { _studentId = value; }
    }

    public int Credits
    {
        get { return _credits; }
        set { _credits = value; }
    }

    public string Scholarship
    {
        get { return _scholarship; }
        set { _scholarship = value; }
    }

    public Student() : base() { }

    public Student(string name, int age, string gender, string hobby, string nationality,
                   string university, int course, string favoriteSubject, double averageGrade)
        : base(name, age, gender, hobby, nationality)
    {
        University = university;
        Course = course;
        FavoriteSubject = favoriteSubject;
        AverageGrade = averageGrade;
    }

    public Student(string name, int age, string gender, string hobby, string nationality,
                   string university, int course, string favoriteSubject, double averageGrade,
                   string group, string studentId, int credits, string scholarship)
        : this(name, age, gender, hobby, nationality, university, course, favoriteSubject, averageGrade)
    {
        Group = group;
        StudentId = studentId;
        Credits = credits;
        Scholarship = scholarship;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Университет: {University}, Курс: {Course}, Любимый предмет: {FavoriteSubject}, Средний балл: {AverageGrade}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, учусь на {Course} курсе в {University}, люблю {Hobby}!");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (студент) обсуждает учебу с {other.Name}.");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} учится по предмету {FavoriteSubject}.");
    }

    public void Study(int hours)
    {
        Console.WriteLine($"{Name} учится {hours} часов по предмету {FavoriteSubject}.");
    }

    public void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен в {University}.");
    }

    public void AttendLecture()
    {
        Console.WriteLine($"{Name} посещает лекцию в группе {Group}.");
    }

    public void SubmitAssignment(string subject)
    {
        Console.WriteLine($"{Name} сдал работу по предмету {subject}.");
    }

    public void RequestScholarship()
    {
        Console.WriteLine($"{Name} подал заявку на стипендию. Текущая стипендия: {Scholarship}.");
    }

    public void ChangeGroup(string newGroup)
    {
        Group = newGroup;
        Console.WriteLine($"{Name} переведен в группу {newGroup}.");
    }

    void IStudent.Study()
    {
        Console.WriteLine($"{Name} изучает {FavoriteSubject} в {University}");
    }

    string IStudent.University
    {
        get { return _university; }
        set { _university = value; }
    }

    int IStudent.Course
    {
        get { return _course; }
        set { _course = value; }
    }

    string IStudent.Group
    {
        get { return _group; }
        set { _group = value; }
    }

    void IStudent.AttendLecture()
    {
        Console.WriteLine($"IStudent: {Name} присутствует на лекции в группе {Group}");
    }
}

public class Employee : Person, IWorker
{
    private string _company;
    private int _salary;
    private string _position;
    private int _experience;
    private string _department;
    private string _employeeId;
    private string _workSchedule;
    private int _vacationDays;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public int Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set { _experience = value; }
    }

    public string Department
    {
        get { return _department; }
        set { _department = value; }
    }

    public string EmployeeId
    {
        get { return _employeeId; }
        set { _employeeId = value; }
    }

    public string WorkSchedule
    {
        get { return _workSchedule; }
        set { _workSchedule = value; }
    }

    public int VacationDays
    {
        get { return _vacationDays; }
        set { _vacationDays = value; }
    }

    public Employee() : base() { }

    public Employee(string name, int age, string gender, string hobby, string nationality,
                    string company, int salary, string position, int experience)
        : base(name, age, gender, hobby, nationality)
    {
        Company = company;
        Salary = salary;
        Position = position;
        Experience = experience;
    }

    public Employee(string name, int age, string gender, string hobby, string nationality,
                    string company, int salary, string position, int experience,
                    string department, string employeeId, string workSchedule, int vacationDays)
        : this(name, age, gender, hobby, nationality, company, salary, position, experience)
    {
        Department = department;
        EmployeeId = employeeId;
        WorkSchedule = workSchedule;
        VacationDays = vacationDays;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, Зарплата: {Salary} руб., Стаж: {Experience} лет";
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (работник) разговаривает с {other.Name}.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} на должности {Position}.");
    }

    public void Work(int hours)
    {
        Console.WriteLine($"{Name} работает {hours} часов в {Company}.");
    }

    public void GetPromotion()
    {
        Salary += 10000;
        Console.WriteLine($"{Name} получил повышение! Новая зарплата: {Salary} руб.");
    }

    public void TakeBreak()
    {
        Console.WriteLine($"{Name} ушел на перерыв в компании {Company}.");
    }

    public void RequestVacation(int days)
    {
        Console.WriteLine($"{Name} запросил отпуск на {days} дней. Доступно: {VacationDays}.");
    }

    public void ChangeDepartment(string newDepartment)
    {
        Department = newDepartment;
        Console.WriteLine($"{Name} переведен в отдел {newDepartment}.");
    }

    void IWorker.Work()
    {
        Console.WriteLine($"{Name} трудится в {Company}");
    }

    string IWorker.Company
    {
        get { return _company; }
        set { _company = value; }
    }

    string IWorker.Department
    {
        get { return _department; }
        set { _department = value; }
    }

    void IWorker.TakeBreak()
    {
        Console.WriteLine($"IWorker: {Name} отдыхает в отделе {Department}");
    }
}

public class Teacher : Employee
{
    private string _subject;
    private string _university;
    private string _academicDegree;
    private int _studentsCount;
    private string _office;
    private string _researchArea;
    private int _publicationsCount;
    private string _teachingStyle;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public string AcademicDegree
    {
        get { return _academicDegree; }
        set { _academicDegree = value; }
    }

    public int StudentsCount
    {
        get { return _studentsCount; }
        set { _studentsCount = value; }
    }

    public string Office
    {
        get { return _office; }
        set { _office = value; }
    }

    public string ResearchArea
    {
        get { return _researchArea; }
        set { _researchArea = value; }
    }

    public int PublicationsCount
    {
        get { return _publicationsCount; }
        set { _publicationsCount = value; }
    }

    public string TeachingStyle
    {
        get { return _teachingStyle; }
        set { _teachingStyle = value; }
    }

    public Teacher() : base() { }

    public Teacher(string name, int age, string gender, string hobby, string nationality,
                   string company, int salary, string position, int experience,
                   string subject, string university, string academicDegree, int studentsCount)
        : base(name, age, gender, hobby, nationality, company, salary, position, experience)
    {
        Subject = subject;
        University = university;
        AcademicDegree = academicDegree;
        StudentsCount = studentsCount;
    }

    public Teacher(string name, int age, string gender, string hobby, string nationality,
                   string company, int salary, string position, int experience,
                   string subject, string university, string academicDegree, int studentsCount,
                   string office, string researchArea, int publicationsCount, string teachingStyle)
        : this(name, age, gender, hobby, nationality, company, salary, position, experience,
               subject, university, academicDegree, studentsCount)
    {
        Office = office;
        ResearchArea = researchArea;
        PublicationsCount = publicationsCount;
        TeachingStyle = teachingStyle;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Университет: {University}, Степень: {AcademicDegree}, Студентов: {StudentsCount}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, преподаю {Subject} в {University} уже {Experience} лет! {AcademicDegree}");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (преподаватель) общается с {other.Name}.");
    }

    public void Teach()
    {
        Console.WriteLine($"{Name} преподает {Subject} в {University}.");
    }

    public void Teach(string topic)
    {
        Console.WriteLine($"{Name} преподает тему \"{topic}\" по предмету {Subject}.");
    }

    public void CheckHomework()
    {
        Console.WriteLine($"{Name} проверяет домашние задания {StudentsCount} студентов.");
    }

    public void PublishResearch()
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} опубликовал исследование в области {ResearchArea}. Всего публикаций: {PublicationsCount}.");
    }

    public void ConsultStudent(string studentName)
    {
        Console.WriteLine($"{Name} консультирует студента {studentName} в кабинете {Office}.");
    }

    public void ChangeOffice(string newOffice)
    {
        Office = newOffice;
        Console.WriteLine($"{Name} переехал в кабинет {newOffice}.");
    }
}

public class WorkingStudent : Student, IWorker
{
    private string _company;
    private string _position;
    private int _salary;
    private int _workHours;
    private string _workDepartment;
    private string _contractType;
    private int _daysOff;
    private string _mentor;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public int Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public int WorkHours
    {
        get { return _workHours; }
        set { _workHours = value; }
    }

    public string WorkDepartment
    {
        get { return _workDepartment; }
        set { _workDepartment = value; }
    }

    public string ContractType
    {
        get { return _contractType; }
        set { _contractType = value; }
    }

    public int DaysOff
    {
        get { return _daysOff; }
        set { _daysOff = value; }
    }

    public string Mentor
    {
        get { return _mentor; }
        set { _mentor = value; }
    }

    public WorkingStudent() : base() { }

    public WorkingStudent(string name, int age, string gender, string hobby, string nationality,
                          string university, int course, string favoriteSubject, double averageGrade,
                          string company, string position, int salary, int workHours)
        : base(name, age, gender, hobby, nationality, university, course, favoriteSubject, averageGrade)
    {
        Company = company;
        Position = position;
        Salary = salary;
        WorkHours = workHours;
    }

    public WorkingStudent(string name, int age, string gender, string hobby, string nationality,
                          string university, int course, string favoriteSubject, double averageGrade,
                          string company, string position, int salary, int workHours,
                          string workDepartment, string contractType, int daysOff, string mentor)
        : this(name, age, gender, hobby, nationality, university, course, favoriteSubject, averageGrade,
               company, position, salary, workHours)
    {
        WorkDepartment = workDepartment;
        ContractType = contractType;
        DaysOff = daysOff;
        Mentor = mentor;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, Зарплата: {Salary} руб., Рабочих часов: {WorkHours}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, учусь в {University} и работаю в {Company}.");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} на должности {Position}.");
    }

    public void Work(int hours)
    {
        Console.WriteLine($"{Name} работает {hours} часов в {Company} и еще успевает учиться.");
    }

    public void TakeBreak()
    {
        Console.WriteLine($"{Name} берет перерыв на работе в {Company}.");
    }

    public void RequestDayOff()
    {
        DaysOff++;
        Console.WriteLine($"{Name} взял выходной. Всего выходных: {DaysOff}.");
    }

    public void MeetMentor()
    {
        Console.WriteLine($"{Name} встречается с наставником {Mentor} в {Company}.");
    }

    public void UpdateContract(string newContract)
    {
        ContractType = newContract;
        Console.WriteLine($"{Name} сменил тип договора на {newContract}.");
    }

    void IWorker.Work()
    {
        Console.WriteLine($"{Name} совмещает учебу и работу в {Company}");
    }

    string IWorker.Company
    {
        get { return _company; }
        set { _company = value; }
    }

    string IWorker.Department
    {
        get { return _workDepartment; }
        set { _workDepartment = value; }
    }

    void IWorker.TakeBreak()
    {
        Console.WriteLine($"IWorker: {Name} отдыхает от работы в {Company}");
    }
}

public class PersonCollection<T> where T : Person
{
    private List<T> _people = new List<T>();

    public void Add(T person)
    {
        _people.Add(person);
    }

    public void Remove(T person)
    {
        _people.Remove(person);
    }

    public void DisplayAll()
    {
        foreach (T person in _people)
        {
            Console.WriteLine(person.GetInfo());
            person.SayHello();
            Console.WriteLine();
        }
    }

    public int Count()
    {
        return _people.Count;
    }
}

public class PersonService
{
    private readonly IPerson _person;

    public PersonService(IPerson person)
    {
        _person = person;
    }

    public void Introduce()
    {
        _person.SayHello();
        Console.WriteLine($"Info: {_person.GetInfo()}");
    }

    public void Notify(string message)
    {
        _person.SendMessage(message);
        Console.WriteLine($"Уведомление отправлено на {_person.Email}, телефон: {_person.Phone}");
    }
}

public class WorkService
{
    private readonly IWorker _worker;

    public WorkService(IWorker worker)
    {
        _worker = worker;
    }

    public void StartWork()
    {
        Console.WriteLine($"Работник из компании {_worker.Company}:");
        _worker.Work();
    }

    public void StartBreak()
    {
        Console.WriteLine($"Отдел: {_worker.Department}");
        _worker.TakeBreak();
    }
}

public class StudyService
{
    private readonly IStudent _student;

    public StudyService(IStudent student)
    {
        _student = student;
    }

    public void StartStudy()
    {
        Console.WriteLine($"Студент из группы {_student.Group}:");
        _student.Study();
    }

    public void GoToLecture()
    {
        Console.WriteLine($"Университет: {_student.University}, курс: {_student.Course}");
        _student.AttendLecture();
    }
}

Student yulia = new Student("Юля", 19, "Женский", "Роблокс", "Русская", "ТИУ", 2, "Математика", 4.5,
                            "Б-21", "ST-001", 60, "Академическая");
Employee vladimir = new Employee("Владимир", 43, "Мужской", "Автозвук", "Русский", "Газпром", 150000, "Начальник отдела", 15,
                                 "Разработка", "EMP-001", "5/2", 28);
Teacher svetlana = new Teacher("Светлана", 47, "Женский", "Чтение", "Русская", "ТИУ", 80000, "Преподаватель", 15, "Математика", "ТИУ", "Кандидат наук", 120,
                               "А-302", "Прикладная математика", 25, "Лекционный");
WorkingStudent maxim = new WorkingStudent("Максим", 22, "Мужской", "Спорт", "Русский", "ТИУ", 4, "Программирование", 4.8, "Яндекс", "Стажер", 40000, 20,
                                          "IT", "Трудовой", 5, "Алексей");
Person sofia = new Person("София", 19, "Женский", "Чтение", "Русская",
                          "sofia@mail.ru", "+7-900-123-45-67", "Тюмень, ул. Ленина 1", "Не замужем");

Console.WriteLine("=== Информация ===");
Console.WriteLine(yulia.GetInfo());
Console.WriteLine(vladimir.GetInfo());
Console.WriteLine(svetlana.GetInfo());
Console.WriteLine(maxim.GetInfo());
Console.WriteLine(sofia.GetInfo());

Console.WriteLine("\n=== Приветствия ===");
List<Person> people = new List<Person> { yulia, vladimir, svetlana, maxim, sofia };
foreach (Person p in people)
{
    p.SayHello();
}

Console.WriteLine("\n=== Перегрузка методов ===");
yulia.Study();
yulia.Study(5);
vladimir.Work();
vladimir.Work(8);
svetlana.Teach();
svetlana.Teach("Интегралы");
maxim.Work();
maxim.Work(6);

Console.WriteLine("\n=== Уникальные методы ===");
yulia.TakeExam();
vladimir.GetPromotion();
svetlana.CheckHomework();

Console.WriteLine("\n=== Новые атрибуты и методы ===");
yulia.AttendLecture();
yulia.SubmitAssignment("Математика");
yulia.RequestScholarship();
yulia.ChangeGroup("Б-22");

vladimir.TakeBreak();
vladimir.RequestVacation(14);
vladimir.ChangeDepartment("Аналитика");

svetlana.PublishResearch();
svetlana.ConsultStudent("Юля");
svetlana.ChangeOffice("А-401");

maxim.TakeBreak();
maxim.RequestDayOff();
maxim.MeetMentor();
maxim.UpdateContract("Гражданско-правовой");

sofia.SendMessage("Всем привет!");
sofia.Call("+7-900-765-43-21");
sofia.UpdateAddress("Тюмень, ул. Мира 10");
sofia.CelebrateBirthday();

Console.WriteLine("\n=== Взаимодействие объектов ===");
yulia.InteractWith(svetlana);
svetlana.InteractWith(yulia);
vladimir.InteractWith(yulia);
maxim.InteractWith(svetlana);

Console.WriteLine("\n=== Явная реализация интерфейса ===");
IPerson personAsInterface = yulia;
Console.WriteLine(personAsInterface.GetInfo());
personAsInterface.SayHello();
personAsInterface.Email = "yulia@tiu.ru";
personAsInterface.Phone = "+7-912-345-67-89";
personAsInterface.SendMessage("Привет через интерфейс IPerson");

IStudent studentAsInterface = yulia;
studentAsInterface.Study();
Console.WriteLine($"Университет через интерфейс: {studentAsInterface.University}");
studentAsInterface.Group = "Б-23";
studentAsInterface.AttendLecture();

IWorker workerAsInterface = vladimir;
workerAsInterface.Work();
Console.WriteLine($"Компания через интерфейс: {workerAsInterface.Company}");
workerAsInterface.Department = "Финансы";
workerAsInterface.TakeBreak();

IWorker workingStudentAsInterface = maxim;
workingStudentAsInterface.Work();
workingStudentAsInterface.Department = "Разработка";
workingStudentAsInterface.TakeBreak();

Console.WriteLine("\n=== Управление зависимостями ===");
PersonService personService = new PersonService(yulia);
personService.Introduce();
personService.Notify("Скоро экзамен!");

Console.WriteLine();

PersonService personService2 = new PersonService(svetlana);
personService2.Introduce();
personService2.Notify("Заседание кафедры");

Console.WriteLine();

WorkService workService = new WorkService(vladimir);
workService.StartWork();
workService.StartBreak();

Console.WriteLine();

WorkService workService2 = new WorkService(maxim);
workService2.StartWork();
workService2.StartBreak();

Console.WriteLine();

StudyService studyService = new StudyService(yulia);
studyService.StartStudy();
studyService.GoToLecture();

Console.WriteLine();

StudyService studyService2 = new StudyService(maxim);
studyService2.StartStudy();
studyService2.GoToLecture();

Console.WriteLine("\n=== Generic класс ===");
PersonCollection<Student> students = new PersonCollection<Student>();
students.Add(yulia);
students.Add(new Student("Олег", 20, "Мужской", "Футбол", "Русский", "ТИУ", 3, "Физика", 4.2));
Console.WriteLine($"Студентов в коллекции: {students.Count()}");
students.DisplayAll();

PersonCollection<Teacher> teachers = new PersonCollection<Teacher>();
teachers.Add(svetlana);
Console.WriteLine($"Преподавателей в коллекции: {teachers.Count()}");
teachers.DisplayAll();

PersonCollection<Employee> employees = new PersonCollection<Employee>();
employees.Add(vladimir);
Console.WriteLine($"Работников в коллекции: {employees.Count()}");
employees.DisplayAll();
```
