### Вариант задания 1


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.

Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Person
{
    private string _name;
    private int _age;
    private string _gender;
    private string _hobby;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Age
    {
        get { return _age; }
        set { _age = value; }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public string Hobby
    {
        get { return _hobby; }
        set { _hobby = value; }
    }

    public Person() { }

    public Person(string name, int age, string gender, string hobby)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Hobby = hobby;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, {Age} лет, {Gender}, Хобби: {Hobby}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}");
    }

    public virtual void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}.");
    }
}

public class Student : Person
{
    private string _university;
    private int _course;
    private string _favoriteSubject;

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public int Course
    {
        get { return _course; }
        set { _course = value; }
    }

    public string FavoriteSubject
    {
        get { return _favoriteSubject; }
        set { _favoriteSubject = value; }
    }

    public Student() : base() { }

    public Student(string name, int age, string gender, string hobby,
                   string university, int course, string favoriteSubject)
        : base(name, age, gender, hobby)
    {
        University = university;
        Course = course;
        FavoriteSubject = favoriteSubject;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, учусь на {Course} курсе в {University}, люблю {Hobby}!");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (студент) обсуждает учебу с {other.Name}.");
    }
}

public class Employee : Person
{
    private string _company;
    private int _salary;
    private string _position;

    public string Company
    {
        get { return _company; }
        set { _company = value; }
    }

    public int Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public Employee() : base() { }

    public Employee(string name, int age, string gender, string hobby,
                    string company, int salary, string position)
        : base(name, age, gender, hobby)
    {
        Company = company;
        Salary = salary;
        Position = position;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Должность: {Position}, Зарплата: {Salary} руб.";
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (работник) разговаривает с {other.Name}.");
    }
}

public class Teacher : Person
{
    private string _subject;
    private int _experience;
    private string _university;
    private string _academicDegree;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set { _experience = value; }
    }

    public string University
    {
        get { return _university; }
        set { _university = value; }
    }

    public string AcademicDegree
    {
        get { return _academicDegree; }
        set { _academicDegree = value; }
    }

    public Teacher() : base() { }

    public Teacher(string name, int age, string gender, string hobby,
                   string subject, int experience, string university, string academicDegree)
        : base(name, age, gender, hobby)
    {
        Subject = subject;
        Experience = experience;
        University = university;
        AcademicDegree = academicDegree;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я {Name}, преподаю в {University} уже {Experience} лет! {AcademicDegree}");
    }

    public override void InteractWith(Person other)
    {
        Console.WriteLine($"{Name} (преподаватель) общается с {other.Name}.");
    }
}

Student yulia = new Student("Юля", 19, "Женский", "Роблокс", "ТИУ", 2, "Математика");
Employee vladimir = new Employee("Владимир", 43, "Мужской", "Автозвук", "Газпром", 150000, "Начальник отдела");
Teacher svetlana = new Teacher("Светлана", 47, "Женский", "Чтение", "Математика", 15, "ТИУ", "Кандидат наук");
Person sofia = new Person("София", 19, "Женский", "Чтение");

Console.WriteLine("=== Информация ===");
Console.WriteLine(yulia.GetInfo());
Console.WriteLine(vladimir.GetInfo());
Console.WriteLine(svetlana.GetInfo());
Console.WriteLine(sofia.GetInfo());

Console.WriteLine("\n=== Приветствия ===");
yulia.SayHello();
vladimir.SayHello();
svetlana.SayHello();
sofia.SayHello();

Console.WriteLine("\n=== Взаимодействие объектов ===");
yulia.InteractWith(svetlana);
svetlana.InteractWith(yulia);
vladimir.InteractWith(yulia);
```
